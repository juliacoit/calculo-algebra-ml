# Semana de 1/fev · SVD e backpropagation

## Aplicação em ML/IA — tópicos latentes com SVD (LSA)

**Objetivo:** achar temas em textos sem rótulo nenhum, só com a SVD de uma matriz de palavras.

**Passo 1.** Junte cerca de 150 textos curtos de 3 temas (por exemplo, 50 de cada) e monte a matriz documento × palavra com `TfidfVectorizer` do scikit-learn.

**Passo 2.** Calcule a SVD truncada com $k = 3$ (`TruncatedSVD` ou `np.linalg.svd` ficando com os 3 primeiros valores singulares).

**Passo 3.** Plote os documentos nas 2 primeiras direções, uma cor por tema (as cores servem só para conferir; o método não as usa).

**Passo 4.** Para cada um dos 3 vetores singulares, liste as 10 palavras de maior peso e dê um nome ao tema.

**O que observar:** a SVD agrupa palavras que aparecem juntas nos mesmos documentos. É a ideia por trás dos embeddings, antes das redes neurais.
