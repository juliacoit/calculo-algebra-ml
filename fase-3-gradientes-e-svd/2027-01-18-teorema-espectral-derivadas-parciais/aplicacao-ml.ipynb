## Aplicação em ML/IA — a superfície de perda de uma regressão

**Objetivo:** ver que a perda de um modelo real é uma forma quadrática e que padronizar as features muda os autovalores dela.

**Passo 1.** Pegue uma única feature de `load_diabetes` e o modelo $\hat{y} = w\cdot x + b$. A perda $L(w, b)$ é uma função de duas variáveis.

**Passo 2.** Plote as curvas de nível de $L$ sobre uma grade de valores de $w$ e $b$.

**Passo 3.** A matriz de segundas derivadas (a Hessiana) dessa perda é $2A^\top A/n$, com $A = [x, 1]$. Calcule-a, confirme que é simétrica e ache seus autovalores.

**Passo 4.** Padronize $x$ (média 0, desvio 1), refaça as curvas e os autovalores e compare a razão entre o maior e o menor.

**O que observar:** sem padronizar, as elipses ficam muito alongadas (razão grande entre autovalores). Padronizar as deixa quase circulares; é por isso que quase todo pipeline de ML padroniza as features.
