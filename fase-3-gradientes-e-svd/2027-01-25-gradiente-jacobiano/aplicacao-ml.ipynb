# Semana de 25/jan · Gradiente, regra da cadeia em várias variáveis e Jacobiano

## Aplicação em ML/IA — treinar com e sem padronização, e o momentum

**Objetivo:** medir como os autovalores da semana anterior controlam a velocidade do treino, e ver o Jacobiano de uma camada.

**Passo 1.** Treine o modelo $\hat{y} = w\cdot x + b$ da semana anterior por gradiente descendente, com e sem padronizar $x$. Conte quantos passos cada versão leva até a perda parar de cair.

**Passo 2.** Implemente o momentum, $v \leftarrow \beta\, v + \nabla L$ e $(w, b) \leftarrow (w, b) - \eta\, v$, com $\beta = 0{,}9$. Compare a trajetória com a da versão simples sobre as curvas de nível.

**Passo 3.** Para uma camada $y = Wx$, mostre no papel que o Jacobiano de $y$ em relação a $x$ é a própria $W$. Confira com `torch.autograd.functional.jacobian`.

**O que observar:** o número de passos acompanha a razão entre os autovalores. O momentum reduz o zigue-zague ao acumular a direção que se repete.
