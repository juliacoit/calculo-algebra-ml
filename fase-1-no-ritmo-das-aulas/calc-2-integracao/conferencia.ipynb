# Cálculo · Integração e aplicações da integral

## Conferência no código — integral como área e como probabilidade

**Passo 1.** Implemente a soma de Riemann pelo ponto médio para $\int_0^1 x^2\,dx$ com $n = 10$, $100$ e $1000$ e compare com o valor exato $1/3$.

**Passo 2.** Calcule à mão a imprópria $\int_0^\infty e^{-x}\,dx$ e aproxime no código truncando em 5, 10 e 20.

**Passo 3.** Verifique numericamente que a densidade normal $\dfrac{1}{\sqrt{2\pi}}\,e^{-x^2/2}$ integra perto de 1 em $[-10, 10]$.

**Passo 4.** Calcule o valor médio de $\operatorname{sen} x$ em $[0, \pi]$ pela integral e compare com a média de 10 000 pontos sorteados no intervalo.
