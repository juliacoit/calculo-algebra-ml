# Álgebra Linear · Autovalores, autovetores e diagonalização

## Aplicação em ML/IA — PCA em embeddings de frases

**Objetivo:** reduzir vetores de 384 dimensões para 2 usando os autovetores da covariância, e ver se o significado sobrevive.

**Passo 1.** Escreva cerca de 40 frases curtas em português sobre 4 temas (por exemplo: répteis, anfíbios, culinária e futebol), 10 por tema.

**Passo 2.** Gere os embeddings com `sentence-transformers` (modelo `paraphrase-multilingual-MiniLM-L12-v2`, que devolve vetores de 384 dimensões).

**Passo 3.** Centralize e calcule a covariância $384\times 384$, como na conferência.

**Passo 4.** Pegue os 2 autovetores de maior autovalor com `np.linalg.eigh` e projete cada frase neles (produto da matriz de dados pelos 2 autovetores).

**Passo 5.** Plote as frases em 2D, uma cor por tema.

**Passo 6.** Calcule a fração da variância explicada: $\dfrac{\lambda_1 + \lambda_2}{\sum_i \lambda_i}$.

**O que observar:** os temas tendem a formar grupos mesmo com 2 de 384 direções. Uma fração explicada baixa mostra que o significado está espalhado por muitas direções.
