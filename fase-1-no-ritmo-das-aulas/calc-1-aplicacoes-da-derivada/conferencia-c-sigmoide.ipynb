# Cálculo · Aplicações da derivada

## Conferência no código C — esboço da sigmoide

**Passo 1.** À mão, faça o estudo completo de $\sigma(x) = \dfrac{1}{1 + e^{-x}}$: limites no infinito, derivada $\sigma' = \sigma(1 - \sigma)$, crescimento e concavidade.

**Passo 2.** Plote $\sigma$ e $\sigma'$ e confira o esboço. Note que $\sigma'$ fica perto de zero para $|x| > 5$: é o "gradiente que some" em redes neurais.
