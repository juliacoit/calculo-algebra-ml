## Conferência no código B — polinômio de Taylor

**Passo 1.** Plote $e^x$ junto com seus polinômios de Taylor de grau 1, 2 e 3 em torno de 0.

**Passo 2.** Meça o erro de cada um em $x = 0{,}1$ e $x = 1$ e compare com o termo de resto.

**Passo 3.** Use o grau 1 para mostrar que $f(x - \eta\, f'(x)) \approx f(x) - \eta\, f'(x)^2$: por isso um passo pequeno contra a derivada diminui $f$.
