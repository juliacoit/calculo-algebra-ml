## Conferência no código — Taylor e esboço da sigmoide

**Passo 1.** Plote $e^x$ junto com seus polinômios de Taylor de grau 1, 2 e 3 em torno de 0 e meça o erro em $x = 0{,}1$ e $x = 1$.

**Passo 2.** Use o grau 1 para mostrar que $f(x - \eta\, f'(x)) \approx f(x) - \eta\, f'(x)^2$: por isso um passo pequeno contra a derivada diminui $f$ (liga com C7).

**Passo 3.** Faça à mão o esboço completo de $\sigma(x) = \dfrac{1}{1 + e^{-x}}$ (limites no infinito, $\sigma' = \sigma(1 - \sigma)$, crescimento, inflexão em 0) e confira no gráfico. Veja que $\sigma'$ fica perto de zero para $|x| > 5$: é o "gradiente que some".
