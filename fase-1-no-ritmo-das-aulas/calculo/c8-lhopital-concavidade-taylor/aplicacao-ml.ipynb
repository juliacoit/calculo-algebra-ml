## Aplicação em ML/IA — entropia: medir a incerteza de um classificador

**Objetivo:** estudar por completo a função que mede incerteza em ML, usando L'Hôpital, concavidade e Taylor.

**Passo 1.** Defina a entropia binária $H(p) = -p\ln p - (1 - p)\ln(1 - p)$ para $p$ em $(0, 1)$.

**Passo 2.** À mão: escreva $p\ln p$ como $\dfrac{\ln p}{1/p}$ e use L'Hôpital para mostrar que o limite é 0 quando $p \to 0^+$. É a justificativa da convenção $0\cdot\ln 0 = 0$ usada em todo código de ML.

**Passo 3.** Calcule $H'(p) = \ln\dfrac{1 - p}{p}$ e $H''(p) = -\dfrac{1}{p(1 - p)}$. Conclua que $H$ é côncava com máximo $\ln 2$ em $p = 1/2$, e faça o esboço completo.

**Passo 4.** Taylor de grau 2 em torno de $1/2$: $H(p) \approx \ln 2 - 2(p - 1/2)^2$. Plote as duas e compare o erro em $p = 0{,}4$ e $p = 0{,}1$.

**Passo 5.** Treine `LogisticRegression` no `load_breast_cancer` do scikit-learn, pegue `predict_proba` no conjunto de teste e calcule $H$ para cada exemplo.

**Passo 6.** Liste os 10 exemplos de maior entropia e confira se são os que o modelo mais erra.

**O que observar:** entropia alta significa modelo indeciso. Escolher para rotular os exemplos de maior entropia é a ideia do aprendizado ativo, útil quando rotular dados custa caro, como identificar espécies em gravações.
