# C6 · Logaritmo e exponencial (Ávila cap. 7)

## Aplicação em ML/IA — log-verossimilhança, entropia cruzada e perplexidade

**Objetivo:** ver por que modelos de linguagem trabalham com logaritmos e calcular a métrica clássica deles.

**Passo 1.** Multiplique 1000 probabilidades iguais a 0,01: o resultado vira 0.0 (estouro por baixo). Some os logaritmos: dá $1000\cdot\ln 0{,}01 \approx -4605{,}17$, sem problema.

**Passo 2.** Monte um corpus de umas 2000 palavras com textos públicos sobre herpetofauna e um modelo de unigramas, $p(w) = \dfrac{\text{contagem de } w}{\text{total}}$. Para palavras que não aparecem, use a suavização de Laplace: $\dfrac{\text{contagem} + 1}{\text{total} + \text{tamanho do vocabulário}}$.

**Passo 3.** Calcule a log-verossimilhança de uma frase nova como a soma de $\ln p(w)$.

**Passo 4.** Calcule a entropia cruzada $H = -\dfrac{1}{n}\sum \ln p(w)$ e a perplexidade $e^H$ para uma frase sobre o tema e uma sobre futebol. A do tema deve ser menor.

**Passo 5.** Plote $-\ln p$ para $p$ em $(0, 1]$. A derivada $-1/p$ explode perto de 0: errar com muita confiança custa caro.

**Passo 6.** Prove à mão que $\ln \sum e^{z_i} = m + \ln \sum e^{z_i - m}$, com $m = \max z_i$, e teste com $z = (1000, 1001)$: a fórmula direta estoura e esta não.

**O que observar:** o logaritmo transforma produto em soma, e isso torna o treino numérico possível. Perplexidade é a exponencial da entropia cruzada: quanto menor, menos o modelo se surpreende com o texto.
