## Aplicação em ML/IA — tempo de espera e máxima verossimilhança

**Objetivo:** calcular média e variância de uma distribuição com integração por partes e ajustar o modelo aos dados.

**Passo 1.** Use a densidade exponencial $f(t) = \lambda e^{-\lambda t}$ para $t \ge 0$, por exemplo o tempo entre dois registros de uma espécie num ponto de escuta. Confira à mão que a imprópria $\int_0^\infty f(t)\,dt$ vale 1.

**Passo 2.** Por partes: $E[T] = \int_0^\infty t\,\lambda e^{-\lambda t}\,dt = 1/\lambda$.

**Passo 3.** Por partes duas vezes: $E[T^2] = 2/\lambda^2$, então a variância é $1/\lambda^2$.

**Passo 4.** Simule 100 000 tempos com `np.random.exponential(1/λ)` e compare média e variância com as fórmulas.

**Passo 5.** Máxima verossimilhança: a log-verossimilhança de $n$ tempos é $\ell(\lambda) = n\ln\lambda - \lambda\sum t_i$. Derive, iguale a zero e mostre que o melhor $\lambda$ é $1/\text{média}$ (usa C6 e C7). Confira nos dados simulados.

**Passo 6.** Calcule $P(T > t) = e^{-\lambda t}$ pela integral e compare com a fração simulada para 3 valores de $t$.

**O que observar:** integrais por partes calculam esperanças, e "ajustar um modelo" é maximizar a log-verossimilhança. A regressão logística da Fase 2 é a mesma ideia com mais parâmetros.
