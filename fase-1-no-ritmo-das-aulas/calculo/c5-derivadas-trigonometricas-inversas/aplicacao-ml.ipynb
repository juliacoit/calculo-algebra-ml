# C5 · Derivadas trigonométricas, inversas e formas indeterminadas (Ávila 6.3 a 6.5)

## Aplicação em ML/IA — sazonalidade de vocalização de anuros

**Objetivo:** ajustar uma senoide a dados mensais por gradiente descendente, derivando cada passo à mão.

**Passo 1.** Simule 36 meses de contagens de registros de vocalização: $y = 20 + 15\operatorname{sen}(2\pi t/12) + 8\cos(2\pi t/12) + \text{ruído}$. (São dados inventados; se tiver uma série real do trabalho, use ela.)

**Passo 2.** Use o modelo $\hat{y} = a\operatorname{sen}(\omega t) + b\cos(\omega t) + c$, com $\omega = 2\pi/12$ fixo, e a perda $L = \text{média de } (\hat{y} - y)^2$.

**Passo 3.** Derive à mão $\dfrac{\partial L}{\partial a}$, $\dfrac{\partial L}{\partial b}$ e $\dfrac{\partial L}{\partial c}$. Cada uma é uma derivada comum, com as outras letras tratadas como constantes.

**Passo 4.** Treine por gradiente descendente e plote os dados com a curva ajustada.

**Passo 5.** Escreva o resultado como $A\operatorname{sen}(\omega t + \varphi)$, com $A = \sqrt{a^2 + b^2}$ e $\varphi$ dado pela arco-tangente (use `np.arctan2(b, a)`), e calcule em que mês fica o pico.

**Passo 6.** Deixe $\omega$ livre: pela cadeia, $\dfrac{\partial \hat{y}}{\partial \omega} = a\,t\cos(\omega t) - b\,t\operatorname{sen}(\omega t)$. Plote $L$ como função de $\omega$ e veja os vários mínimos locais: o ponto de partida decide onde o treino para.

**O que observar:** codificar mês, hora ou dia da semana como seno e cosseno é padrão em ML para tempo cíclico. A arco-tangente recupera a fase, que é o que interessa biologicamente: a época do pico.
