## Aplicação em ML/IA — derivada numérica e o erro de h

**Objetivo:** entender a ferramenta usada para verificar se um gradiente está certo (o gradient check do Marco 3).

**Passo 1.** Use a sigmoide $\sigma(x) = \dfrac{1}{1 + e^{-x}}$ em $x = 0{,}5$. A derivada exata é $\sigma(x)(1 - \sigma(x))$ (você prova isso em C3).

**Passo 2.** Calcule a diferença progressiva $\dfrac{f(x + h) - f(x)}{h}$ e a central $\dfrac{f(x + h) - f(x - h)}{2h}$ para $h = 10^{-1}, 10^{-2}, \dots, 10^{-12}$.

**Passo 3.** Plote o erro de cada uma contra $h$ em escala log-log. O gráfico tem forma de V: para $h$ grande domina o erro da aproximação; para $h$ pequeno, o arredondamento do computador.

**Passo 4.** Ache o melhor $h$ de cada fórmula (fica perto de $10^{-8}$ na progressiva e de $10^{-5}$ na central).

**Passo 5.** Na ReLU em $x = 0$, as razões laterais dão 0 e 1. Confira com `torch` qual valor a biblioteca usa como "derivada" em 0.

**O que observar:** a derivada numérica é a definição de derivada com $h$ pequeno, mas não pequeno demais. Num ponto sem derivada, como a ReLU em 0, as bibliotecas simplesmente escolhem um valor.
