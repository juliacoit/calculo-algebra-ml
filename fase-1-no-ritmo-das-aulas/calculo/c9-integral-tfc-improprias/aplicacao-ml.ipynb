# C9 · Integral definida, Teorema Fundamental do Cálculo e impróprias (Ávila cap. 10)

## Aplicação em ML/IA — um classificador bayesiano em uma dimensão

**Objetivo:** usar integrais para transformar densidades em probabilidades e medir o erro de um classificador como uma área.

**Passo 1.** Simule o comprimento de 200 indivíduos de duas espécies, cada espécie com distribuição normal de média e desvio diferentes.

**Passo 2.** Estime média e desvio de cada espécie e escreva as duas densidades.

**Passo 3.** Calcule $P(a < X < b)$ para uma espécie pela soma de Riemann e compare com a fração dos dados simulados nesse intervalo.

**Passo 4.** Classifique cada indivíduo pela espécie de maior densidade e ache o ponto de fronteira em que as duas densidades se cruzam.

**Passo 5.** Calcule o erro esperado como a área das caudas que passam da fronteira (duas integrais) e compare com o erro medido nos dados.

**O que observar:** a integral liga um modelo de probabilidade ao número que importa, a chance de errar. É a base de Probabilidade e Estatística A.
