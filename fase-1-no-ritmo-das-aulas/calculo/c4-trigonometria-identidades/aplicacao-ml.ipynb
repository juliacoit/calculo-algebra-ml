# C4 · Funções trigonométricas, identidades e limite fundamental (Ávila 6.1 e 6.2)

## Aplicação em ML/IA — a codificação posicional do Transformer

**Objetivo:** ver como seno e cosseno dizem ao modelo em que posição da frase cada palavra está, e por que as fórmulas de soma de arcos são o motivo de isso funcionar.

**Passo 1.** Para $d = 16$ e posições 0 a 49, monte a matriz $PE$ de $50\times 16$ com $PE(pos, 2i) = \operatorname{sen}\left(pos/10000^{2i/d}\right)$ e $PE(pos, 2i + 1) = \cos\left(pos/10000^{2i/d}\right)$.

**Passo 2.** Plote a matriz como mapa de calor: as primeiras colunas oscilam rápido e as últimas, devagar.

**Passo 3.** À mão: o período da coluna $2i$ é $2\pi\cdot 10000^{2i/d}$. Calcule para $i = 0$ e $i = 7$.

**Passo 4.** Pelas fórmulas de soma de arcos, mostre que o par $(\operatorname{sen}(\omega(p + k)), \cos(\omega(p + k)))$ é uma matriz de rotação que depende só de $k$ aplicada a $(\operatorname{sen}\omega p, \cos\omega p)$. Confira para $p = 3$ e $k = 5$.

**Passo 5.** Calcule o produto interno $PE(p)\cdot PE(q)$ para vários pares e plote contra $|p - q|$. Prove à mão que ele é a soma de $\cos(\omega_i(p - q))$ e, portanto, só depende da distância.

**O que observar:** a fórmula de soma de arcos transforma "andar k posições" numa rotação. É a mesma ideia do RoPE, a codificação posicional dos LLMs atuais.
