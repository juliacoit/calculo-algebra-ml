# C4 · Funções trigonométricas, identidades e limite fundamental (Ávila 6.1 e 6.2)

## Conferência no código — identidades e o limite fundamental

**Passo 1.** Plote as seis funções em $[-2\pi, 2\pi]$, marcando período e assíntotas.

**Passo 2.** Teste $\operatorname{sen}(a + b) = \operatorname{sen} a\cos b + \cos a\operatorname{sen} b$ e $\cos(a + b) = \cos a\cos b - \operatorname{sen} a\operatorname{sen} b$ em 1000 pares aleatórios.

**Passo 3.** Tabele $\dfrac{\operatorname{sen} x}{x}$ e $\dfrac{1 - \cos x}{x}$ para $x = 10^{-1}, \dots, 10^{-6}$.

**Passo 4.** Monte a matriz de rotação $R(\theta) = \begin{bmatrix} \cos\theta & -\operatorname{sen}\theta \\ \operatorname{sen}\theta & \cos\theta \end{bmatrix}$ e confira que $R(a)\,R(b) = R(a + b)$. Multiplique à mão e veja que as entradas são as fórmulas de soma de arcos (liga com AL2).
