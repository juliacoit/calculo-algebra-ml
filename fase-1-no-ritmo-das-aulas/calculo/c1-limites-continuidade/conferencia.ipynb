## Conferência no código — limites pela tabela

**Passo 1.** Para um limite feito à mão, monte a tabela de $f(a - h)$ e $f(a + h)$ com $h = 10^{-1}, 10^{-2}, \dots, 10^{-6}$ e confira os dois lados (o mesmo teste da folha de consulta: $x = a \pm 0{,}001$).

**Passo 2.** Confira com `sympy.limit`, inclusive os laterais (`dir='+'` e `dir='-'`).

**Passo 3.** Plote uma função racional com suas assíntotas desenhadas em tracejado.

**Passo 4.** Use o Teorema do Valor Intermediário na bisseção: ache a raiz de $x^3 - x - 1$ em $[1, 2]$ com erro menor que $10^{-6}$, dividindo o intervalo ao meio a cada passo.
