# C1 · Limites e continuidade (Ávila 4.2; revisão)

## Aplicação em ML/IA — a temperatura de um modelo de linguagem

**Objetivo:** ver que o "criativo" e o "determinístico" de um LLM são dois limites da mesma função.

**Passo 1.** Suponha que o modelo deu os escores $z = (2{,}0;\ 1{,}0;\ 0{,}5;\ -1{,}0)$ para 4 palavras candidatas.

**Passo 2.** Implemente $\operatorname{softmax}(z/T)$, em que a probabilidade da palavra $i$ é $e^{z_i/T}$ dividido pela soma dos $e^{z_j/T}$, e calcule para $T = 5;\ 2;\ 1;\ 0{,}5;\ 0{,}1;\ 0{,}01$.

**Passo 3.** À mão: mostre que, quando $T \to 0^+$, a probabilidade da maior palavra tende a 1 e as outras a 0 (divida tudo por $e^{z_{\max}/T}$ e use o limite de $e^{-c/T}$ com $c > 0$). Mostre que, quando $T \to \infty$, todas tendem a $1/4$.

**Passo 4.** Plote as 4 probabilidades contra $T$, com $T$ em escala logarítmica.

**Passo 5.** Sorteie 20 "próximas palavras" com $T = 0{,}2$ e com $T = 2$ e compare a variedade.

**Passo 6.** Calcule os limites de $\pm\infty$ da sigmoide e da tanh e verifique que a ReLU, $\max(0, x)$, é contínua em 0.

**O que observar:** a temperatura que você ajusta num chatbot é esse parâmetro $T$. Temperatura baixa se aproxima do limite $T \to 0$ (sempre a palavra mais provável); temperatura alta, do limite $T \to \infty$ (sorteio uniforme).
