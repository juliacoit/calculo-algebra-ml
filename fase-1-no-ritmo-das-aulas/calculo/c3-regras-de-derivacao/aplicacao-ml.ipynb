## Aplicação em ML/IA — backpropagation de um neurônio

**Objetivo:** ver que treinar uma rede é aplicar a regra da cadeia, e que o PyTorch faz a mesma conta que você.

**Passo 1.** Monte o neurônio $z = wx + b$, $p = \sigma(z)$ e a perda $L = (p - y)^2$. Desenhe no papel o grafo $x \to z \to p \to L$.

**Passo 2.** À mão: prove $\sigma'(z) = \sigma(z)(1 - \sigma(z))$ pela regra do quociente e escreva $\dfrac{\partial L}{\partial w} = 2(p - y)\cdot\sigma'(z)\cdot x$ pela cadeia (e $\dfrac{\partial L}{\partial b}$, igual sem o $x$).

**Passo 3.** Implemente a ida (forward) e a volta (backward) em Python puro para $x = 2$, $y = 1$, $w = 0{,}5$, $b = 0$.

**Passo 4.** Refaça com `torch.tensor(..., requires_grad=True)` e `L.backward()` e confira que os gradientes são iguais.

**Passo 5.** Treine $w$ e $b$ por 100 passos de gradiente descendente em 50 pontos simulados e plote a perda.

**Passo 6.** A função logit, $\ln\dfrac{p}{1 - p}$, é a inversa da sigmoide. Calcule sua derivada pela regra da inversa e confira que dá $\dfrac{1}{p(1 - p)}$.

**O que observar:** backpropagation é a regra da cadeia aplicada de trás para frente no grafo. O logit é o valor que o modelo produz antes da sigmoide, e a regra da inversa liga os dois.
