## Aplicação em ML/IA — treinar um modelo de um parâmetro

**Objetivo:** ver que "treinar" é minimizar uma função de uma variável, e que as derivadas limitam a taxa de aprendizado.

**Passo 1.** Gere 50 pontos com $y = 3x + \text{ruído}$.

**Passo 2.** Use o modelo $\hat{y} = w\cdot x$ e a perda $L(w) = \text{média de } (y - w\cdot x)^2$. Note que $L$ é uma parábola em $w$.

**Passo 3.** Ache o $w$ ótimo à mão fazendo $L'(w) = 0$.

**Passo 4.** Treine por gradiente descendente usando a $L'(w)$ que você derivou. Plote $L(w)$ e marque cada passo sobre a curva.

**Passo 5.** Calcule $L''(w)$ e confirme que o treino diverge quando $\eta$ passa de $2/L''$.

**Passo 6.** Pelo TVM, $|\sigma(a) - \sigma(b)| \le M\cdot|a - b|$, com $M$ o máximo de $\sigma'$. Ache $M = 1/4$ à mão e teste a desigualdade em 10 000 pares aleatórios.

**O que observar:** o $w$ do gradiente descendente converge para o mesmo $w$ da conta à mão, e o limite de $\eta$ sai da 2ª derivada. O TVM garante que a sigmoide nunca amplia diferenças mais que 4 vezes; é o tipo de limite usado para analisar a estabilidade do treino.
