## Conferência no código — gradiente descendente em uma variável

**Passo 1.** Para $f(x) = (x - 3)^2 + 1$, ache o mínimo à mão igualando $f'(x)$ a zero.

**Passo 2.** Implemente a repetição $x \leftarrow x - \eta\, f'(x)$, partindo de $x = -5$, por 50 passos.

**Passo 3.** Teste $\eta = 0{,}01$; $0{,}1$; $0{,}9$ e $1{,}1$ e plote $x$ a cada passo. Com $1{,}1$ o processo diverge; explique por quê usando $f''(x) = 2$.

**Passo 4.** Repita com $f(x) = x^4 - 3x^2 + x$, que tem dois mínimos locais: o ponto de partida decide onde o processo para.
