## Aplicação em ML/IA — montar curvas com funções de ativação transladadas

**Objetivo:** ver que uma rede neural de uma camada é uma soma de cópias transladadas e esticadas de uma mesma função, e que isso basta para aproximar uma curva.

**Passo 1.** Plote $\sigma(k(x - a))$ para $k = 1$, $5$ e $50$ e $a = -1$, $0$ e $2$. Identifique a translação ($a$) e a mudança de escala ($k$): com $k$ grande, a sigmoide vira um degrau em $x = a$.

**Passo 2.** Monte um "degrau de largura finita" com $h(x) = \sigma(k(x - a)) - \sigma(k(x - b))$ e explique à mão por que, com $k$ grande, $h$ vale perto de 1 entre $a$ e $b$ e perto de 0 fora.

**Passo 3.** Aproxime a parábola $f(x) = x^2$ em $[-2, 2]$ somando 10 desses degraus, cada um com a altura de $f$ no seu centro. Plote e meça o erro máximo com 10, 20 e 50 degraus.

**Passo 4.** Repita com a ReLU: combinações de $\operatorname{ReLU}(x - a)$ dão funções lineares por partes. Escolha à mão 5 pontos de quebra e as inclinações que fazem a soma passar pela parábola nesses pontos.

**Passo 5.** Plote $\sigma$ e a logit, $\ln\dfrac{p}{1 - p}$, com a reta $y = x$. Anote domínio e imagem de cada uma ($\mathbb{R} \to (0, 1)$ e o contrário) e confira $\sigma(\operatorname{logit}(p)) = p$.

**Passo 6.** Lei de Zipf (uma hipérbole): conte as palavras de um texto longo em português, ordene por frequência e plote a frequência contra a posição $r$. Compare com $C/r$ e veja que, em escala log-log, vira uma reta.

**O que observar:** cada neurônio de uma camada oculta é a mesma função transladada e esticada, e a saída da rede é a soma deles. Esse é o teorema da aproximação universal visto com as ferramentas do capítulo 3. A lei de Zipf explica por que poucas palavras dominam qualquer texto, o que afeta como vocabulários de modelos de linguagem são montados.
