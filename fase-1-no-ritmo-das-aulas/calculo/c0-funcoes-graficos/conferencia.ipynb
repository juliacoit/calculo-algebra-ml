# C0 · Funções e gráficos (Ávila cap. 3; revisão)

## Conferência no código — gráficos, translações e inversas

**Passo 1.** Plote $f(x) = x^2$ junto com $f(x - 2)$, $f(x) + 1$, $f(2x)$ e $3f(x)$ e confira, para cada uma, a regra de translação ou de escala que você aplicou à mão.

**Passo 2.** Para uma parábola de exercício, complete o quadrado à mão e marque o vértice no gráfico.

**Passo 3.** Plote $f$ e a inversa que você calculou, junto com a reta $y = x$: os gráficos são reflexos um do outro. Confira $f(f^{-1}(x)) = x$ em 100 pontos.

**Passo 4.** Plote $x^2 + y^2 = 25$ e veja que ela não é gráfico de uma função; separe em dois ramos, $y = \sqrt{25 - x^2}$ e $y = -\sqrt{25 - x^2}$.
