# AL6 · Gram-Schmidt, complemento ortogonal e projeção (Marco 1)

## Aplicação em ML/IA — busca semântica, a etapa de recuperação de um RAG (Marco 1)

**Objetivo:** dada uma pergunta, achar os trechos mais parecidos de um corpus usando só produto interno, norma e projeção.

**Passo 1.** Monte um corpus de uns 50 parágrafos de textos públicos sobre répteis e anfíbios brasileiros, um parágrafo por item de uma lista.

**Passo 2.** Gere os embeddings, empilhe numa matriz $E$ (uma linha por parágrafo) e normalize cada linha para norma 1.

**Passo 3.** Gere o embedding da pergunta, normalize e calcule `E @ q`: um único produto dá o cosseno com todos os parágrafos.

**Passo 4.** Ordene com `np.argsort` e devolva os 3 maiores.

**Passo 5.** Teste 5 perguntas e anote em quantas o trecho certo veio em 1º lugar.

**Passo 6.** Tire de $q$ a sua projeção sobre o embedding da palavra "cobra" ($q - \langle q, c\rangle c$, com $c$ normalizado) e repita a busca. Os resultados devem se afastar desse tema: é o complemento ortogonal em ação.

**O que observar:** é exatamente a recuperação de um chatbot RAG. Normalizar antes transforma o cosseno em produto interno, que se calcula para o corpus inteiro numa só multiplicação.
