# AL3 · Autovalores e autovetores

## Aplicação em ML/IA — PageRank

**Objetivo:** ranquear páginas de um site pelos links, calculando um autovetor.

**Passo 1.** Invente um site com 6 páginas sobre anfíbios e desenhe os links entre elas (quem aponta para quem).

**Passo 2.** Monte a matriz de transição $M$: $M_{ij} = 1/(\text{número de links que saem de } j)$ se $j$ aponta para $i$, senão 0. Cada coluna soma 1.

**Passo 3.** Acrescente o amortecimento: $G = 0{,}85\cdot M + 0{,}15\cdot(1/6)$ em todas as entradas.

**Passo 4.** Aplique o método das potências a partir de $r = (1/6, \dots, 1/6)$ até a mudança ficar abaixo de $10^{-8}$.

**Passo 5.** Compare com o autovetor de $\lambda = 1$ dado por `np.linalg.eig(G)`, normalizado para somar 1, e ordene as páginas.

**Passo 6.** Remova um link, rode de novo e veja quem sobe e quem desce.

**O que observar:** o ranking é o autovetor de autovalor 1. Esse autovalor existe porque as colunas de $G$ somam 1, e o método das potências converge porque os outros autovalores têm módulo menor.
