# Álgebra Linear · Transformações lineares e suas matrizes

## Aplicação em ML/IA — uma camada linear separando duas classes

**Objetivo:** ver que uma camada de rede neural, $W\cdot x + b$, é uma transformação linear seguida de translação, e que o núcleo de $W$ decide que informação se perde.

**Passo 1.** Gere duas classes de pontos 2D com `make_blobs` (scikit-learn) e plote, uma cor por classe.

**Passo 2.** Escolha à mão um vetor $W$ de $1\times 2$ e um número $b$ para que a reta $W\cdot x + b = 0$ separe as classes. Classifique cada ponto pelo sinal de $W\cdot x + b$ e calcule a acurácia.

**Passo 3.** Mude $W$ e $b$ à mão e veja a reta girar ($W$ muda a direção) e se deslocar ($b$ translada).

**Passo 4.** Agora use $W$ de $2\times 2$, plote os pontos transformados `W @ X` e veja a camada remexer o plano.

**Passo 5.** Troque $W$ por uma matriz de posto 1: todos os pontos caem numa reta. Calcule o núcleo e mostre que pontos que diferem por um vetor do núcleo viram o mesmo ponto.

**O que observar:** uma camada com núcleo não trivial apaga informação para sempre. É o mesmo que acontece quando uma camada estreita comprime um embedding.
