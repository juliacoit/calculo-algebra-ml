# AL2 · Matriz de uma transformação, composição e mudança de base

## Aplicação em ML/IA — por que redes neurais precisam de não linearidade

**Objetivo:** ver que empilhar camadas lineares é compor TLs, o que dá só outra TL, e que uma função não linear entre elas muda isso.

**Passo 1.** Gere duas classes com `make_blobs` e classifique pelo sinal de $W\cdot x + b$ com $W$ e $b$ escolhidos à mão. Mude $W$ (a reta gira) e $b$ (a reta se desloca).

**Passo 2.** Monte duas camadas sem ativação, $W_2(W_1x + b_1) + b_2$. Calcule $W = W_2W_1$ e $b = W_2b_1 + b_2$ e confirme que uma única camada dá as mesmas saídas.

**Passo 3.** Use o XOR: os pontos $(0,0)$ e $(1,1)$ na classe 0 e $(0,1)$ e $(1,0)$ na classe 1. Teste uma grade de valores de $W$ e $b$ e veja que nenhuma reta acerta os 4 pontos (no máximo 3).

**Passo 4.** Coloque a ReLU, $\max(0, \cdot)$, entre as camadas, com $W_1 = \begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix}$, $b_1 = (0, -1)$ e $W_2 = [1, -2]$. Calcule à mão a saída nos 4 pontos e confira no código: dá 0, 1, 1, 0.

**Passo 5.** Plote os 4 pontos depois de $W_1$ e da ReLU: no novo plano, uma reta já os separa.

**O que observar:** sem não linearidade, uma rede de 100 camadas é uma só matriz. A ReLU deforma o espaço, e é isso que permite separar o que nenhuma reta separa.
