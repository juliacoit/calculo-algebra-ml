# AL1 · Transformações lineares: definição, núcleo e imagem

## Conferência no código — núcleo, imagem e dimensão

**Passo 1.** Defina $T$ pela matriz de um exercício feito à mão e use `sympy.Matrix` para obter `nullspace()`, `columnspace()` e `rank()`. Compare com suas bases.

**Passo 2.** Confira o teorema: número de colunas $= \dim N(T) + \dim \operatorname{Im}(T)$.

**Passo 3.** Teste a linearidade com 100 pares de vetores aleatórios e mostre que $T(x) = Ax + b$ falha quando $b \neq 0$ (basta olhar $T(0)$).

**Passo 4.** Escreva a derivação de $P_3$ em $P_2$ como uma matriz $3\times 4$ agindo nos coeficientes; aplique a $1 + 2x + 3x^2 + 4x^3$ e ache o núcleo (os polinômios constantes).
