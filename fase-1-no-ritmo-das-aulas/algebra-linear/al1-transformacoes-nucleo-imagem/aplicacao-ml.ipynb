# AL1 · Transformações lineares: definição, núcleo e imagem

## Aplicação em ML/IA — features redundantes

**Objetivo:** ler um conjunto de dados como uma transformação linear e ver o que o núcleo significa para um modelo.

**Passo 1.** Carregue `load_diabetes` do scikit-learn: $X$ tem 442 pacientes e 10 features. Pense em $X$ como uma TL de $\mathbb{R}^{10}$ em $\mathbb{R}^{442}$ que leva pesos $w$ em previsões $Xw$.

**Passo 2.** Calcule o posto ($\dim \operatorname{Im}$) e deduza $\dim N$ pelo teorema do núcleo e da imagem.

**Passo 3.** Acrescente uma 11ª coluna igual a $2 \times \text{coluna 1} + \text{coluna 3}$. Confirme que o posto não sobe e ache à mão o vetor $n$ que gera o núcleo.

**Passo 4.** Pegue qualquer $w$ e mostre que $X\cdot w$ e $X\cdot(w + t\cdot n)$ dão as mesmas previsões para $t = 1$, $10$ e $1000$.

**Passo 5.** Calcule o número de condição de $X^\top X$ (`np.linalg.cond`) antes e depois da coluna extra.

**O que observar:** um núcleo não trivial faz o modelo ter infinitos pesos igualmente bons, e os coeficientes deixam de ter significado. A regularização da Fase 2 (semana de 4/jan) resolve isso.
