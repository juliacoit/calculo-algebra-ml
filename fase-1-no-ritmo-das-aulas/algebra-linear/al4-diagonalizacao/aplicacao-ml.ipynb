# AL4 · Diagonalização

## Aplicação em ML/IA — PCA em embeddings de frases

**Objetivo:** diagonalizar a matriz de covariância para reduzir vetores de 384 dimensões a 2, e ver se o significado sobrevive.

**Passo 1.** Aquecimento em 2D: gere 200 pontos alongados com `np.random.randn(200, 2) @ np.array([[3, 0], [1, 0.5]])`, centralize, calcule $C = X^\top X/199$ e diagonalize $C = PDP^\top$. Plote os pontos e os autovetores: o de maior autovalor aponta para onde a nuvem é mais comprida.

**Passo 2.** Escreva cerca de 40 frases curtas em português sobre 4 temas (por exemplo: répteis, anfíbios, culinária e futebol), 10 por tema.

**Passo 3.** Gere os embeddings com `sentence-transformers` (modelo `paraphrase-multilingual-MiniLM-L12-v2`, vetores de 384 dimensões).

**Passo 4.** Centralize, calcule a covariância $384\times 384$ e pegue os 2 autovetores de maior autovalor com `np.linalg.eigh`.

**Passo 5.** Projete cada frase nesses 2 autovetores e plote em 2D, uma cor por tema.

**Passo 6.** Calcule a fração da variância explicada: $\dfrac{\lambda_1 + \lambda_2}{\sum_i \lambda_i}$.

**O que observar:** os temas tendem a formar grupos mesmo usando 2 de 384 direções. $P^\top = P^{-1}$ porque a covariância é simétrica. Uma fração explicada baixa mostra que o significado está espalhado por muitas direções.
