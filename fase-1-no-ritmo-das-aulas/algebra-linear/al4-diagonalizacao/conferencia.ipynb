# AL4 · Diagonalização

## Conferência no código — Fibonacci por diagonalização

**Passo 1.** Escreva a recorrência como $(F_{n+1}, F_n) = A\cdot(F_n, F_{n-1})$, com $A = \begin{bmatrix} 1 & 1 \\ 1 & 0 \end{bmatrix}$.

**Passo 2.** Diagonalize $A$ à mão: os autovalores são $\dfrac{1 \pm \sqrt{5}}{2}$.

**Passo 3.** Calcule $F_{30}$ por $PD^{30}P^{-1}$ e compare com um laço simples ($F_{30} = 832\,040$).

**Passo 4.** Tente diagonalizar $\begin{bmatrix} 1 & 1 \\ 0 & 1 \end{bmatrix}$: `np.linalg.eig` devolve dois autovetores praticamente iguais, e $P$ fica quase singular. Confirme à mão que ela não é diagonalizável.
