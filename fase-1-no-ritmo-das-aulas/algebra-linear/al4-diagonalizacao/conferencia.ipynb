# Álgebra Linear · Autovalores, autovetores e diagonalização

## Conferência no código — a direção em que os dados mais variam

**Passo 1.** Gere 200 pontos alongados: `X = np.random.randn(200, 2) @ np.array([[3, 0], [1, 0.5]])`.

**Passo 2.** Subtraia a média de cada coluna e calcule a matriz de covariância `C = X.T @ X / 199`.

**Passo 3.** Resolva $\det(C - \lambda I) = 0$ à mão (é uma equação do 2º grau) e ache os autovetores. Confira com `np.linalg.eigh(C)`.

**Passo 4.** Plote os pontos e os dois autovetores saindo da origem: o de maior autovalor aponta para onde a nuvem é mais comprida.

**Passo 5.** Verifique numericamente $C = PDP^\top$. Aqui $P^\top = P^{-1}$ porque $C$ é simétrica (isso volta na Fase 3).
