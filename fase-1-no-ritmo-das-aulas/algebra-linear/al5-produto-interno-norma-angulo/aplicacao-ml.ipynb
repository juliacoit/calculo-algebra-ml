# AL5 · Produto interno, norma e ângulo

## Aplicação em ML/IA — vizinhos mais próximos (k-NN)

**Objetivo:** ver que escolher o produto interno é escolher o que "parecido" quer dizer para um classificador.

**Passo 1.** Carregue `load_iris` (150 flores, 4 medidas em cm) e separe 70% para treino e 30% para teste.

**Passo 2.** Implemente o k-NN com $k = 5$: para cada flor de teste, ache as 5 mais próximas pela distância euclidiana e vote a espécie. Calcule a acurácia.

**Passo 3.** Multiplique uma das medidas por 100 (como se ela estivesse numa unidade 100 vezes menor) e repita: a acurácia cai, porque essa coordenada domina a distância.

**Passo 4.** Padronize as 4 medidas (média 0, desvio 1) e recupere a acurácia. Isso equivale a usar o produto interno $u^\top M v$ com $M$ diagonal.

**Passo 5.** Com os embeddings de frases de AL4, compare o k-NN por distância euclidiana e por cosseno, antes e depois de normalizar os vetores.

**O que observar:** com vetores de norma 1, $\lVert u - v\rVert^2 = 2 - 2\cos\theta$, então distância e cosseno dão o mesmo ranking. Prove essa igualdade à mão a partir de $\lVert u - v\rVert^2 = \langle u - v, u - v\rangle$.
