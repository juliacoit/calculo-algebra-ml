# AL5 · Produto interno, norma e ângulo

## Conferência no código — normas, ângulos e Cauchy-Schwarz

**Passo 1.** Confira as normas e os ângulos feitos à mão com `np.linalg.norm` e `np.arccos`.

**Passo 2.** Teste $|\langle u, v\rangle| \le \lVert u\rVert\,\lVert v\rVert$ em 1000 pares aleatórios e ache o par mais próximo da igualdade (vetores quase paralelos).

**Passo 3.** Calcule $\langle f, g\rangle = \int_0^1 f\,g\,dx$ para dois polinômios com `sympy.integrate` e compare com a conta à mão.

**Passo 4.** Use $\langle u, v\rangle = u^\top M v$ com $M = \begin{bmatrix} 1 & 2 \\ 2 & 1 \end{bmatrix}$ e ache um $u$ com $\langle u, u\rangle < 0$: $M$ não é definida positiva, então não define produto interno.
