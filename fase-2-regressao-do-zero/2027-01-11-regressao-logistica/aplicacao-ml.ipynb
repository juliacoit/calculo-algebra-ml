# Semana de 11/jan · Marco 2: regressão logística

## Aplicação em ML/IA — classificador de sentimento do zero

**Passo 1.** Use um conjunto pequeno de frases rotuladas como positivas ou negativas e transforme cada frase num vetor de contagens de palavras (bag-of-words).

**Passo 2.** Implemente em NumPy a previsão $p$, a perda $L$ e o gradiente $(p - y)\,x$.

**Passo 3.** Treine por gradiente descendente e plote a perda a cada época.

**Passo 4.** Compare a acurácia com a de `LogisticRegression` do scikit-learn no mesmo conjunto.

**Passo 5.** Liste as 10 palavras com maior peso positivo e as 10 com maior peso negativo e confira se fazem sentido.

**O que observar:** o gradiente $(p - y)\,x$ tem a mesma forma do gradiente da regressão linear, $(\hat{y} - y)\,x$. Só a função de saída mudou.
