# Semana de 4/jan · Projeção e mínimos quadrados

## Aplicação em ML/IA — regressão linear de verdade e regularização

**Objetivo:** aplicar mínimos quadrados a dados reais com várias features e ver como o termo $\lambda I$ resolve o problema da semana anterior.

**Passo 1.** Com `load_diabetes`, separe 80% dos pacientes para treino e 20% para teste.

**Passo 2.** Acrescente uma coluna de 1s e resolva as equações normais com `np.linalg.lstsq`.

**Passo 3.** Compare os coeficientes com os de `LinearRegression` do scikit-learn: devem ser iguais.

**Passo 4.** Calcule o erro quadrático médio no teste.

**Passo 5.** Volte à matriz com a coluna redundante e resolva a versão regularizada $(X^\top X + \lambda I)\,w = X^\top y$ para $\lambda = 0{,}1$, $1$ e $10$. Compare o erro de teste de cada uma.

**O que observar:** somar $\lambda I$ soma $\lambda$ a todos os autovalores de $X^\top X$. Nenhum fica zero, então a matriz vira invertível; é a regressão Ridge.
