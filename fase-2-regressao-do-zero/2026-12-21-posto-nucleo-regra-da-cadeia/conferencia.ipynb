# Semana de 21/dez · Posto, núcleo e regra da cadeia

## Conferência no código — posto e derivada

**Passo 1.** Monte uma matriz $5\times 3$ em que a 3ª coluna é a soma das outras duas. Calcule o posto à mão e confira com `np.linalg.matrix_rank`.

**Passo 2.** Ache à mão um vetor $v$ do núcleo e confirme que `A @ v` dá zero.

**Passo 3.** Derive $g(w) = \sigma(3w + 1)$ pela regra da cadeia e compare com a diferença central $\dfrac{g(w + h) - g(w - h)}{2h}$, com $h = 10^{-5}$.
