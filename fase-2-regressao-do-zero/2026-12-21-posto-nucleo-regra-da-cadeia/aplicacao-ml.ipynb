# Semana de 21/dez · Posto, núcleo e regra da cadeia

## Aplicação em ML/IA — features redundantes e o autograd

**Objetivo:** ver o que o posto diz sobre um conjunto de dados, e confirmar que o PyTorch faz a mesma regra da cadeia que você faz à mão.

**Passo 1.** Carregue `load_diabetes` do scikit-learn (442 pacientes, 10 features) e calcule o posto da matriz $X$.

**Passo 2.** Acrescente uma 11ª coluna igual a $2 \times \text{coluna 1} + \text{coluna 3}$. Confirme que o posto não sobe e ache o vetor do núcleo que essa coluna cria.

**Passo 3.** Calcule o número de condição de $X^\top X$ (`np.linalg.cond`) antes e depois da coluna extra e veja o salto.

**Passo 4.** Para um neurônio $p = \sigma(w\cdot x + b)$, escreva $\dfrac{\partial p}{\partial w}$ à mão. Depois calcule com `torch.autograd` e confira que os números batem.

**O que observar:** uma feature que é combinação das outras cria um núcleo não trivial, e o modelo passa a ter infinitas soluções igualmente boas. O autograd é só a regra da cadeia automatizada.
