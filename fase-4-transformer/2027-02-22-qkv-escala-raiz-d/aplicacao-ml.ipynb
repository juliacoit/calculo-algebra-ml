## Aplicação em ML/IA — a matriz de atenção de um modelo real

**Objetivo:** abrir um modelo pré-treinado e reconhecer nele a matriz que você montou à mão.

**Passo 1.** Carregue o GPT-2 pela biblioteca `transformers` com `output_attentions=True`.

**Passo 2.** Rode uma frase curta em inglês e pegue a matriz de atenção de uma cabeça de uma camada. Ela tem formato tokens $\times$ tokens.

**Passo 3.** Confira as duas propriedades da semana: cada linha soma 1 (softmax) e tudo acima da diagonal é zero (máscara causal).

**Passo 4.** Plote a matriz como mapa de calor, com os tokens nos eixos, e anote para qual token anterior cada palavra mais olha.

**Passo 5.** Compare 3 ou 4 cabeças diferentes da mesma camada.

**O que observar:** a matriz do modelo real tem exatamente a forma e as propriedades da sua. Cada cabeça aprende um padrão diferente de produtos internos.
