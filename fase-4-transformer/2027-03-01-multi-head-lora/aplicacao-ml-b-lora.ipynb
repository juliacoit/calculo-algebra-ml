# Semana de 1/mar · Marco 4 e posto baixo

## Aplicação em ML/IA — por que o LoRA funciona

**Objetivo:** medir, numa matriz de pesos real, quanto uma aproximação de posto baixo economiza e quanto perde.

**Passo 1.** Conte os parâmetros: para $d = 768$ e $r = 8$, uma matriz cheia tem $d^2 = 589\,824$ números e o par $B$, $A$ tem $2\cdot d\cdot r = 12\,288$, cerca de 2%.

**Passo 2.** Pegue uma matriz de pesos de atenção do GPT-2, calcule a SVD e plote os valores singulares em ordem.

**Passo 3.** Aproxime a matriz com posto 8, 32 e 128 pela SVD truncada e meça o erro relativo de cada uma.

**O que observar:** os valores singulares caem rápido, então pouca informação mora nas direções finais. O LoRA aposta que o ajuste fino precisa de uma mudança de posto baixo nos pesos, não de uma matriz inteira nova.
