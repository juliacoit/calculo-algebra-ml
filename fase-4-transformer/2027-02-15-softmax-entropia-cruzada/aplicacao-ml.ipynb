# Semana de 15/fev · Softmax, Jacobiano e entropia cruzada

## Aplicação em ML/IA — um modelo de próxima letra

**Objetivo:** treinar um modelo de linguagem minúsculo usando só a softmax e o gradiente $s - y$ que você derivou.

**Passo 1.** Pegue uma lista de uns 1000 nomes (por exemplo, nomes científicos de espécies ou nomes próprios) e trate cada nome como uma sequência de letras, com um símbolo de início e de fim.

**Passo 2.** Represente a letra atual como um vetor one-hot $x$ de 27 posições e crie uma matriz $W$ de $27\times 27$ com valores aleatórios pequenos.

**Passo 3.** Calcule os logits $z = W^\top x$, a softmax $s$ e a perda de entropia cruzada contra a letra seguinte de verdade.

**Passo 4.** Atualize $W$ pelo gradiente, usando $\dfrac{\partial L}{\partial z} = s - y$, até a perda média parar de cair.

**Passo 5.** Gere 10 nomes novos sorteando uma letra de cada vez a partir de $s$.

**O que observar:** não há nenhuma derivada além da que você fez à mão. O modelo aprende quais letras costumam vir depois de quais.
